
## 1) Introducción a los números primos

Un **número primo** es un número mayor que 1 que solo tiene dos divisores: él mismo y 1.

Ejemplos: 2, 3, 5, 7, 11, 13, 17, 19...

En criptografía, los números primos grandes se usan para generar llaves seguras.


```python

# @title Generar lista de números primos pequeños
def es_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

primos = [n for n in range(2, 50) if es_primo(n)]
print("Primos menores a 50:", primos)
```


## 2) Generación de llaves pública y privada (versión didáctica)

Usaremos un **RSA simplificado**:

1. Elegimos dos números primos `p` y `q`.
2. Calculamos `n = p * q`.
3. Calculamos `φ(n) = (p-1)*(q-1)`.
4. Elegimos `e` tal que sea coprimo con `φ(n)` (llave pública).
5. Calculamos `d` tal que `(d * e) % φ(n) = 1` (llave privada).


```python

# @title Generar llaves públicas y privadas simples
import math

# Elegimos primos pequeños para la demostración
p = 17
q = 23
n = p * q
phi = (p - 1) * (q - 1)

# Elegimos e
e = 3
while math.gcd(e, phi) != 1:
    e += 2

# Calculamos d
def modinv(a, m):
    for x in range(1, m):
        if (a * x) % m == 1:
            return x
    return None

d = modinv(e, phi)

print(f"p={p}, q={q}, n={n}, φ(n)={phi}")
print(f"Clave pública: (e={e}, n={n})")
print(f"Clave privada: (d={d}, n={n})")
```


## 3) Cifrado y Descifrado

- **Cifrar:** \( C = M^e \mod n \)
- **Descifrar:** \( M = C^d \mod n \)

Donde:
- `M` es el mensaje como número
- `C` es el texto cifrado como número


```python

# @title Ejemplo de cifrado/descifrado con números
# Mensaje como número (ej: 'HOLA' → números pequeños en este demo)
M = 42

# Cifrar
C = pow(M, e, n)

# Descifrar
M_descifrado = pow(C, d, n)

print(f"Mensaje original: {M}")
print(f"Cifrado: {C}")
print(f"Descifrado: {M_descifrado}")
```
